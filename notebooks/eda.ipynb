Loading Dataset and Inspect

```python
import pandas as pd

df = pd.read_csv(
    "../data/raw/train_FD001.txt",
    sep=r"\s+",
    header=None
)

df.head()
df.shape
df.describe()
```

Rename columns with proper names

```python
columns = ['unit', 'cycle', 'op1', 'op2', 'op3']

for i in range(1, 22):
    columns.append(f'sensor_{i}')

df.columns = columns
df.head()
```

Removing Empty Columns

```python
df = df.dropna(axis=1, how='all')
```

Calculate RUL (RUL = max_cycle - current_cycle)

```python
max_cycle = df.groupby('unit')['cycle'].max()

df = df.merge(max_cycle.rename('max_cycle'), on='unit')  

df['RUL'] = df['max_cycle'] - df['cycle']
df.head()
```

Simple Visualization

```python
import matplotlib.pyplot as plt

engine_1 = df[df['unit'] == 1]

plt.plot(engine_1['cycle'], engine_1['sensor_2'])
plt.xlabel("Cycle")
plt.ylabel("Sensor 2")
plt.show()
```

Prepare Features

```python
sensor_columns = [col for col in df.columns if 'sensor' in col]

X = df[sensor_columns]
y = df['RUL']
```

Train First Model

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2
)

model = RandomForestRegressor()

model.fit(X_train, y_train)

preds = model.predict(X_test)

print(mean_absolute_error(y_test, preds))
```

Save Model

```python
import joblib

joblib.dump(model, "../models/rf_model.pkl")
```
